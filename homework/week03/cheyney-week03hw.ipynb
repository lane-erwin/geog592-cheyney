```python
# students code
import csv

filename =  "tabela1613.csv"

with open(filename, "r", encoding= "utf-8") as file:
    reader = csv.reader(file)

    for x in range(5):
        next(reader)



    rows = list(reader)

print(len(rows), "total rows")
print(rows[:5])
```

```python
columns = [
    "Nível",
    "Cód.",
    "Município",
    "Total",
    "Unidade_Total",
    "Açaí",
    "Unidade_Açaí",

]
data = []

for row in rows:
    if len(row) == 7:
        data.append(dict(zip(columns, row)))

print(data[0].keys())
print(data[:3])
```

```python
para = []

for row in data:
    if row["Município"].endswith("(PA)"):
        para.append(row)

print("Number of municipalities in Para:", len(para))
print(para[:5])
```

```python
### keep only the columns I need which are Cod and Acai
cleaned = []

for row in para:
    cleaned.append({
        "Cód.": row["Cód."],
        "Açaí": row["Açaí"],
    })

print(cleaned[:5])
```

```python
## look for what doesnt look like a number, then you will get an array
# ask to go through each row of acai and to send into that array whatever is not a number. name it "not_numbers" and then print it out.
# whatever is not a number, append it to the array "not_numbers"
# create an empty list
nv =[]
for row in cleaned:
    try:
        float(row['Açaí'])  
    except (ValueError, TypeError):
        nv.append(row['Açaí'])
print(nv)
```

```python
for row in cleaned:
    if row["Açaí"] is None or row["Açaí"] == '-':
        row["Açaí"] = None
    else:
        try:
            row["Açaí"] = float(row["Açaí"])
        except ValueError:
            row["Açaí"] = None    
print(cleaned[:10])
```

```python
cleaned = [
    row for row in cleaned
    if row["Açaí"] is not None
]

print("Number of municipalities in Para with Açaí data:", len(cleaned))
```

```python
output_file = "para_acai_cleaned.csv"

with open(output_file, "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["Cód.", "Açaí"])
    writer.writeheader()
    writer.writerows(cleaned)

print("Cleaned file is saved as", output_file)
```
