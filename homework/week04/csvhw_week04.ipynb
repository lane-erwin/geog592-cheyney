```python
import csv

filename = "IDN_ADM2_vulnerability.csv"

with open(filename, "r") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

print("Records found:", len(rows))
print("Column names:", reader.fieldnames)
```

```python
for row in rows[:5]:  
    print(row)
```

```python
print(type(rows[0]["total_pop"]))
print(type(rows[0]["dependency_ratio"]))
print(type(rows[0]["total_pop_rural"]))
```

```python
for column in reader.fieldnames:
    missing = 0
    for row in rows:
        if row[column] == "":
            missing += 1
    print(column, "missing values:", missing)
```

```python
columns_to_keep = [
    "ADM2_PCODE",
    "total_pop",
    "total_pop_rural",
    "dependency_ratio",
    "dependency_ratio_rural"
]
```

```python
cleaned = []

for row in rows:
    new_row = {}

    for column in columns_to_keep:
        new_row [column] = row[column]
    cleaned.append(new_row)

print(cleaned[:5])

print(len(cleaned))
```

```python
for row in cleaned:
    row["total_pop"] = int(row["total_pop"])
    row["total_pop_rural"] = int(row["total_pop_rural"])
    row["dependency_ratio"] = float(row["dependency_ratio"])
    row["dependency_ratio_rural"] = float(row["dependency_ratio_rural"])

print(type(cleaned[0]["total_pop"]))
print(type(cleaned[0]["total_pop_rural"]))
print(type(cleaned[0]["dependency_ratio"]))
print(type(cleaned[0]["dependency_ratio_rural"]))
```

```python
output_filename = "IDN_vulnerability_cleaned.csv"

with open(output_filename, "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=columns_to_keep)
    writer.writeheader()
    writer.writerows(cleaned)
```
